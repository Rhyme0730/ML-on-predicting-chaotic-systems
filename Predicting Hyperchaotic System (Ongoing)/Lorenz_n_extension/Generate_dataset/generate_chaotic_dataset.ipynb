```python
import numpy as np
from scipy.integrate import solve_ivp
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

## System Dynamics

```python
def noise_Lorenz(t, state, a, b, c):
    noise_x = 0.01*(2*np.random.rand()-1)
    noise_y = 0.01*(2*np.random.rand()-1)
    noise_z = 0.01*(2*np.random.rand()-1)
    x, y, z = state
    dstate = np.zeros_like(state)
    dstate[0] = a*(y-x) + noise_x
    dstate[1] = c*x - y - x*z + noise_y
    dstate[2] = x*y - b*z + noise_z

    return dstate
```

```python
def Lorenz(t, state, a, b, c):
    x, y, z = state
    dstate = np.zeros_like(state)
    dstate[0] = a*(y-x)
    dstate[1] = c*x - y - x*z
    dstate[2] = x*y - b*z

    return dstate
```

```python
def Rossler(t, state, a, b, c):
    x, y, z = state
    dstate = np.zeros_like(state)
    dstate[0] = - y - z
    dstate[1] = x + a*y
    dstate[2] = z*x - c*z + b

    return dstate
```

```python
def Chen(t, state, a, b, c):
    x, y, z = state
    dstate = np.zeros_like(state)
    dstate[0] = a*(y-x)
    dstate[1] = (c-a)*x + c*y -x*z
    dstate[2] = x*y - b*z

    return dstate
```

```python
def Qi(t, state, a, b, c):
    x, y, z = state
    dstate = np.zeros_like(state)
    dstate[0] = a*(y-x) + y*z
    dstate[1] = c*x - y - x*z
    dstate[2] = x*y - b*z

    return dstate
```

## Using RK45 to solve ODE

```python
tspan = [0, 50]
teval = np.arange(tspan[0], tspan[1], 0.01)
# initial_states = [17.67715816276679, 12.931379185960404, 43.91404334248268]
initial_states = [1, 1, 1]
method = 'RK23'

data_Lorenz = solve_ivp(fun=Lorenz, t_span=tspan, y0=initial_states, t_eval=teval, method=method, args=(10, 8/3, 28))
data_Lorenz = np.vstack((data_Lorenz.t, data_Lorenz.y)).T

data_Rossler = solve_ivp(fun=Rossler, t_span=tspan, y0=initial_states, t_eval=teval, method=method, args=(0.2, 0.2, 5.7))
data_Rossler = np.vstack((data_Rossler.t, data_Rossler.y)).T

data_Chen = solve_ivp(fun=Chen, t_span=tspan, y0=initial_states, t_eval=teval, method=method, args=(35, 3, 28))
data_Chen = np.vstack((data_Chen.t, data_Chen.y)).T

data_Qi = solve_ivp(fun=Qi, t_span=tspan, y0=initial_states, t_eval=teval, method=method, args=(35, 7, 25))
data_Qi = np.vstack((data_Qi.t, data_Qi.y)).T


data_noise_Lorenz = solve_ivp(fun=noise_Lorenz, t_span=tspan, y0=initial_states, t_eval=teval, method=method, args=(10, 8/3, 28))
data_noise_Lorenz = np.vstack((data_noise_Lorenz.t, data_noise_Lorenz.y)).T


print(data_Lorenz.shape, data_Rossler.shape, data_Chen.shape, data_Qi.shape)
```

## Plot the dataset

```python
def plot_data(data, ax, name):
    N = data.shape[0]
    ax.set_title(name)
    ax.set_xlabel('$x$')
    ax.set_ylabel('$y$')
    ax.set_zlabel('$z$')
    ax.grid(False)

    # ax.plot(data[:, 1], data[:, 2], data[:, 3], cmap='viridis')
    for i in range(N-1):
        ax.plot(data[i:i+2, 1], data[i:i+2, 2], data[i:i+2, 3], color=plt.cm.magma(255*i//N), lw=1.0)
        
def plot_all_data():
    fig = plt.figure(figsize=(18, 5))
    ax1 = fig.add_subplot(141, projection='3d')
    plot_data(data=data_Lorenz, ax=ax1, name='Lorenz system')
    ax2 = fig.add_subplot(142, projection='3d')
    plot_data(data=data_Rossler, ax=ax2, name='Rössler system')
    ax3 = fig.add_subplot(143, projection='3d')
    plot_data(data=data_Chen, ax=ax3, name='Chen system')
    ax4 = fig.add_subplot(144, projection='3d')
    plot_data(data=data_Qi, ax=ax4, name='Qi system')
    plt.show()
```
